```python
import os, requests
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Connecting to the dataset
# printing the columns to the specified Cincinnati data set 
# Print the number of rows and columns

r = requests.get(
    "https://data.cincinnati-oh.gov/api/v3/views/rvmt-pkmq/query.json", # API End Point for each data set
    headers={"X-App-Token": os.environ["SOCRATA_APP_TOKEN"]},
    params={"$limit": 1000, "$offset": 1000}, # Also use "$limit": 1000
)
df = pd.DataFrame(r.json())
print(df.columns.tolist())
print(len(df))
```

```python
# Showing the catalog of available data sets from Cincinnati open data

cat = requests.get(
    "https://api.us.socrata.com/api/catalog/v1",
    params={"domains": "data.cincinnati-oh.gov", "only": "dataset", "limit": 100},
).json()

for r in cat["results"]:
    print(r["resource"]["id"], "|", r["resource"]["name"])
```
